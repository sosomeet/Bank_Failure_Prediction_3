```python
import pandas as pd
import os

BASE_DIR = "."
FILES_IN_ORDER = [
    "results_O.csv",
    "results_F.csv",
    "results_OF.csv",
    "results_OFR.csv",
]

# 1. OFR을 기준 컬럼으로 사용
ofr_path = os.path.join(BASE_DIR, "results_OFR.csv")
df_ofr = pd.read_csv(ofr_path)
base_columns = list(df_ofr.columns)

print(f"[INFO] OFR base columns ({len(base_columns)}): {base_columns}")

dfs = []

# 2. 나머지 파일들을 OFR 컬럼 기준으로 reindex
for fname in FILES_IN_ORDER:
    path = os.path.join(BASE_DIR, fname)
    df = pd.read_csv(path)

    # OFR 기준 컬럼으로 정렬, 없는 건 NaN
    df_aligned = df.reindex(columns=base_columns)

    dfs.append(df_aligned)
    print(f"[ALIGN] {fname}: shape={df_aligned.shape}")

# 3. 순서 유지해서 병합
merged_df = pd.concat(dfs, axis=0, ignore_index=True)

out_path = os.path.join(BASE_DIR, "results_ALL.csv")
merged_df.to_csv(out_path, index=False)

print(f"[DONE] merged shape = {merged_df.shape}")
print(f"[SAVED] {out_path}")
```
