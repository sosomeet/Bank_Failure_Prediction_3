```python
import os
import pandas as pd

# =========================
# Config
# =========================
F_PATH  = "./results_F_over_100_datasets.csv"
OF_PATH = "./results_OF_over_100_datasets.csv"

OUT_DIR = "./mean"
OUT_F   = os.path.join(OUT_DIR, "mean_results_F.csv")
OUT_OF  = os.path.join(OUT_DIR, "mean_results_OF.csv")

# "Score" = 평가지표 컬럼들만 평균
SCORE_COLS = ["AUROC", "AUPRC", "F1", "Brier", "ECE"]

# =========================
# Helpers
# =========================
def mean_scores_horizontal(df: pd.DataFrame, score_cols) -> pd.DataFrame:
    # 실제 존재하는 컬럼만 사용(없으면 에러로 잡아줌)
    missing = [c for c in score_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing score columns: {missing}\nAvailable columns: {list(df.columns)}")

    df_score = df[score_cols].apply(pd.to_numeric, errors="coerce")
    mean_series = df_score.mean(axis=0, skipna=True)

    # 가로 1행 형태
    return pd.DataFrame([mean_series.values], columns=mean_series.index)


def main():
    for p in [F_PATH, OF_PATH]:
        if not os.path.exists(p):
            raise FileNotFoundError(f"File not found: {p}")

    os.makedirs(OUT_DIR, exist_ok=True)

    df_F  = pd.read_csv(F_PATH)
    df_OF = pd.read_csv(OF_PATH)

    mean_F  = mean_scores_horizontal(df_F,  SCORE_COLS)
    mean_OF = mean_scores_horizontal(df_OF, SCORE_COLS)

    mean_F.to_csv(OUT_F, index=False)
    mean_OF.to_csv(OUT_OF, index=False)

    print(f"Saved: {OUT_F}  shape={mean_F.shape}")
    print(f"Saved: {OUT_OF} shape={mean_OF.shape}")


if __name__ == "__main__":
    main()
```
