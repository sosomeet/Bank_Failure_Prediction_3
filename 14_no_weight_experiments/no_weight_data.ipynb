```python
import pandas as pd
from itertools import combinations
from pathlib import Path

# ============================================================
# Config
# ============================================================
INPUT_PATH = Path("./training_data_standardization.csv")
OUTPUT_PATH = Path("./training_data_standardization_with_pairwise_products.csv")

LABEL_COL = "label"

# ============================================================
# Load
# ============================================================
df = pd.read_csv(INPUT_PATH)

# ============================================================
# Feature columns
# - label 제외
# ============================================================
feature_cols = [col for col in df.columns if col != LABEL_COL]

print(f"전체 컬럼 수: {len(df.columns)}")
print(f"feature 수(label 제외): {len(feature_cols)}")
print(f"예상 조합 수 nC2: {len(feature_cols) * (len(feature_cols) - 1) // 2}")

# ============================================================
# Pairwise product features 생성
# - 각 feature 조합의 곱
# ============================================================
pair_list = list(combinations(feature_cols, 2))

for col1, col2 in pair_list:
    new_col = f"{col1}__mul__{col2}"
    df[new_col] = df[col1] * df[col2]

# ============================================================
# Check
# ============================================================
added_cols = [f"{col1}__mul__{col2}" for col1, col2 in pair_list]

print(f"실제 추가된 조합 컬럼 수: {len(added_cols)}")
print(f"최종 컬럼 수: {df.shape[1]}")
print("앞부분 생성 컬럼 예시:")
print(added_cols[:10])

# ============================================================
# Save
# ============================================================
df.to_csv(OUTPUT_PATH, index=False, encoding="utf-8-sig")

print(f"저장 완료: {OUTPUT_PATH.resolve()}")
```
